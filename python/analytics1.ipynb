```python

df_2018 = pd.read_excel("../data/raw/archive/2018 Cyber Crime Stats.xlsx")
df_2019 = pd.read_excel("../data/raw/archive/2019 Cyber Crime Stats.xlsx")

df_2018.head()
```

```python
df_2018["year"] = 2018
```

```python
df_2019["year"] = 2019
```

```python
df_2018.rename(columns={"Terrorist Recuit- ment": "Terrorist Recruitment"}, inplace=True)
df_2019.rename(columns={"Terrorist Recuit- ment": "Terrorist Recruitment"}, inplace=True)
```

```python
df_2018.columns = df_2018.columns.str.strip()
df_2019.columns = df_2019.columns.str.strip()
```

```python
df_2018["City"] = df_2018["City"].str.strip()
df_2019["City"] = df_2019["City"].str.strip()
```

```python
df = pd.concat([df_2018 ,df_2019] , ignore_index = True)
```

```python

df["year"].value_counts()
```

```python
df.to_csv("../data/cleaned/cybercrime_cleaned.csv", index=False)
```

```python
df.head()
```

```python
yearly_totals = df.groupby("year")["Total"].sum()
print(yearly_totals)
```

```python
change = yearly_totals[2019] - yearly_totals[2018]
pct_change = (change / yearly_totals[2018]) * 100
print(f"Change: {change} cases ({pct_change:.1f}%)")
```

```python
import matplotlib.pyplot as plt

yearly_totals.plot(kind="bar", color=["steelblue", "darkorange"])
plt.title("Total Cybercrime Cases: 2018 vs 2019")
plt.ylabel("Total Cases")
plt.xlabel("Year")
plt.xticks(rotation=0)
plt.show()
```

```python
# Top 10 cities overall (both years combined)
top_cities = df.groupby("City")["Total"].sum().sort_values(ascending=False).head(10)
print(top_cities)

top_cities.plot(kind="barh", color="crimson")
plt.title("Top 10 Cities by Total Cybercrime Cases (2018+2019)")
plt.xlabel("Total Cases")
plt.gca().invert_yaxis()   # highest at top
plt.show()
bottom_cities = df.groupby("City")["Total"].sum().sort_values().head(10)
print(bottom_cities)
```

```python
pivot = df.pivot(index="City", columns="year", values="Total")
pivot.columns = ["2018", "2019"]
pivot["Change"] = pivot["2019"] - pivot["2018"]
pivot["% Change"] = (pivot["Change"] / pivot["2018"]) * 100

pivot_sorted = pivot.sort_values("Change", ascending=False)
print(pivot)
print("Biggest increases:\n", pivot_sorted.head(5))
print("\nBiggest decreases:\n", pivot_sorted.tail(5))
```

```python
motive_cols = df.columns[1:-3]
motive_totals = df[motive_cols].sum().sort_values(ascending = False)
print(motive_totals)

motive_totals.head(10).plot(kind="bar", color="teal")
plt.title("Top 10 Cybercrime Motives (Total Cases)")
plt.ylabel("Total Cases")
plt.xticks(rotation=75)
plt.show()
```

```python
print(df["Total"].describe())
print(df.groupby("year")["Total"].describe())
```

```python
Q1 = df["Total"].quantile(0.25)
Q3 = df["Total"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df["Total"] < lower_bound) | (df["Total"] > upper_bound)]
print(outliers[["City", "year", "Total"]].sort_values("Total", ascending=False))
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(df["Total"], vert=True)
plt.title("Cybercrime Total Cases — Outlier Detection")
plt.ylabel("Total Cases")
plt.show()
```

```python
df.boxplot(column="Total", by="year", figsize=(6,4))
plt.title("Cybercrime Distribution by Year")
plt.suptitle("")  # removes the default ugly subtitle
plt.ylabel("Total Cases")
plt.show()
```

```python
Q1_rate = df["Crime Rate"].quantile(0.25)
Q3_rate = df["Crime Rate"].quantile(0.75)
IQR_rate = Q3_rate - Q1_rate

rate_outliers = df[(df["Crime Rate"] < Q1_rate - 1.5*IQR_rate) | (df["Crime Rate"] > Q3_rate + 1.5*IQR_rate)]
print(rate_outliers[["City", "year", "Crime Rate"]])
```
